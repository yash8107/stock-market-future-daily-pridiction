```python
# %% [code] Core Configuration
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Bidirectional
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from sklearn.preprocessing import RobustScaler
import matplotlib.pyplot as plt
import yfinance as yf

# %% [code] Enhanced Data Acquisition & Preprocessing
def load_financial_data(ticker="PRAJIND.NS", start_date="2020-01-01", end_date="2025-04-27"):
    """Load and engineer financial time-series data"""
    df = yf.download(ticker, start=start_date, end=end_date)

    # Feature Engineering
    df['Daily_Return'] = df['Close'].pct_change()
    df['Volatility_7D'] = df['Daily_Return'].rolling(7).std()
    df['EMA_20'] = df['Close'].ewm(span=20, adjust=False).mean()

    # Target Engineering (Multi-horizon prediction)
    forecast_horizon = 1  # Days ahead to predict
    df['Future_High'] = df['High'].shift(-forecast_horizon)
    df['Future_Low'] = df['Low'].shift(-forecast_horizon)

    df.dropna(inplace=True)
    return df

# %% [code] Temporal Feature Scaling
def temporal_scaling(dataframe):
    """Robust scaling for financial data (resistant to outliers)"""
    scaler = RobustScaler()
    features = ['Open', 'High', 'Low', 'Close', 'Volume',
                'Daily_Return', 'Volatility_7D', 'EMA_20']
    scaled_features = scaler.fit_transform(dataframe[features])
    return scaled_features, scaler

# %% [code] 3D Sequence Generator
def create_sequences(data, seq_length=60):
    """Convert time series into supervised learning format"""
    X, y_high, y_low = [], [], []
    for i in range(len(data)-seq_length-1):
        X.append(data[i:i+seq_length])
        y_high.append(data[i+seq_length, 1])  # High price index
        y_low.append(data[i+seq_length, 2])    # Low price index
    return np.array(X), np.array(y_high), np.array(y_low)

# %% [code] Advanced LSTM Architecture
def create_lstm_model(input_shape):
    """Updated LSTM architecture with proper loss specification"""
    model = Sequential([
        tf.keras.layers.Input(shape=input_shape),  # Explicit input layer
        Bidirectional(LSTM(128, return_sequences=True)),
        Dropout(0.3),
        Bidirectional(LSTM(64)),
        Dropout(0.2),
        Dense(32, activation='relu'),
        Dense(2)
    ])

    model.compile(optimizer='adam',
                 loss=tf.keras.losses.Huber(),  # Correct Huber loss syntax
                 metrics=['mae', 'mse'])
    return model

# %% [code] Execution Pipeline
if __name__ == "__main__":
    # Data Preparation
    df = load_financial_data()
    scaled_data, scaler = temporal_scaling(df)

    # Sequence Creation
    SEQ_LENGTH = 60  # 3 months lookback
    X, y_high, y_low = create_sequences(scaled_data, SEQ_LENGTH)
    y = np.column_stack((y_high, y_low))

    # Temporal Train-Test Split
    split = int(0.85 * len(X))
    X_train, X_test = X[:split], X[split:]
    y_train, y_test = y[:split], y[split:]

    # Model Training
    model = create_lstm_model((X_train.shape[1], X_train.shape[2]))

    callbacks = [
        EarlyStopping(patience=15, restore_best_weights=True),
        ModelCheckpoint('best_lstm.h5', save_best_only=True)
    ]

    history = model.fit(X_train, y_train,
                       epochs=100,
                       batch_size=32,
                       validation_split=0.2,
                       callbacks=callbacks,
                       verbose=1)

    # %% [code] Inverse Scaling for Interpretation
    def inverse_scale(predictions, scaler, feature_pos=1):
        """Convert scaled predictions to actual prices"""
        dummy = np.zeros((len(predictions), scaled_data.shape[1]))
        dummy[:, feature_pos] = predictions
        return scaler.inverse_transform(dummy)[:, feature_pos]

    # Prediction and Inverse Scaling
    y_pred = model.predict(X_test)
    actual_high = inverse_scale(y_test[:, 0], scaler, 1)
    pred_high = inverse_scale(y_pred[:, 0], scaler, 1)
    actual_low = inverse_scale(y_test[:, 1], scaler, 2)
    pred_low = inverse_scale(y_pred[:, 1], scaler, 2)

    # %% [code] Advanced Analytics
    plt.figure(figsize=(18, 9))

    # High Prices
    plt.subplot(2, 1, 1)
    plt.plot(actual_high, label='Actual High', alpha=0.7)
    plt.plot(pred_high, label='Predicted High', linestyle='--')
    plt.title('High Price Prediction Performance')
    plt.ylabel('Price')
    plt.legend()

    # Low Prices
    plt.subplot(2, 1, 2)
    plt.plot(actual_low, label='Actual Low', alpha=0.7)
    plt.plot(pred_low, label='Predicted Low', linestyle='--')
    plt.title('Low Price Prediction Performance')
    plt.ylabel('Price')
    plt.legend()

    plt.tight_layout()
    plt.show()

    # %% [code] Quantitative Evaluation
    from sklearn.metrics import mean_absolute_percentage_error

    print("\n=== High Price Evaluation ===")
    print(f"MAE: ${np.mean(np.abs(actual_high - pred_high)):.2f}")
    print(f"MAPE: {mean_absolute_percentage_error(actual_high, pred_high)*100:.2f}%")

    print("\n=== Low Price Evaluation ===")
    print(f"MAE: ${np.mean(np.abs(actual_low - pred_low)):.2f}")
    print(f"MAPE: {mean_absolute_percentage_error(actual_low, pred_low)*100:.2f}%")
```
